```python
import numpy as np
import json,pickle
from pymatgen.core import Structure

def symm_structure(structure):
    from pymatgen.symmetry.analyzer import SpacegroupAnalyzer
    # Save the symm structures
    sga = SpacegroupAnalyzer(structure, symprec=1e-3)
    refined_struc = sga.get_refined_structure()
    sga = SpacegroupAnalyzer(refined_struc,symprec=0.01)
    strc_symmetry = sga.get_symmetrized_structure()
    return strc_symmetry

# Load the data
data = np.load('mp_ids_thermal_conductivity.npz')
kc = data['kc']
kp = data['kp']
klat = data['klat']
mp_ids = data['mp_ids']
print(len(mp_ids))

# Load the structures
with open('Structures_not_sym.json') as f:
    pymatgen_dict_structures_not_sym = json.load(f)

pymatgen_structures_not_sym = [Structure.from_dict(s) for s in pymatgen_dict_structures_not_sym]
pymatgen_structures = [symm_structure(s) for s in pymatgen_structures_not_sym]
```
